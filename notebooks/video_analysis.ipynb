# SIH 2026: CCTV Video Source Analysis & Exploration
This notebook inspects input CCTV surveillance video streams, extracts metadata, and visualizes raw frames.

```python
import os
import sys
import cv2
import matplotlib.pyplot as plt

# Add project root to sys.path
sys.path.insert(0, os.path.abspath('..'))
from src.video.reader import VideoReader
```

## 1. Inspect Available CCTV Surveillance Videos

```python
video_dir = '../videos'
videos = [f for f in os.listdir(video_dir) if f.endswith(('.mp4', '.avi', '.mov'))]
print(f"Found {len(videos)} video files in {video_dir}:\n")

for v in videos:
    vpath = os.path.join(video_dir, v)
    reader = VideoReader(vpath)
    info = reader.get_info()
    print(f"🎥 {v}:")
    print(f"   Resolution:   {info['width']} x {info['height']}")
    print(f"   FPS:          {info['fps']}")
    print(f"   Total Frames: {info['total_frames']} ({info['duration_sec']} sec)\n")
    reader.release()
```

## 2. Sample Frame Visualization

```python
sample_video = os.path.join(video_dir, videos[0])
reader = VideoReader(sample_video)

fig, axes = plt.subplots(1, 4, figsize=(20, 5))
for i, (idx, frame) in enumerate(reader.iter_frames()):
    if i >= 4:
        break
    rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    axes[i].imshow(rgb)
    axes[i].set_title(f"Frame #{idx}")
    axes[i].axis('off')

plt.tight_layout()
plt.show()
reader.release()
```
