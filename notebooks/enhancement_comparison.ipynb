# SIH 2026: Enhancement Model Comparison & Benchmark
This notebook demonstrates side-by-side visual comparisons of raw CCTV footage vs Zero-DCE++, RVRT, Real-ESRGAN, and the Unified Adaptive Pipeline.

```python
import os
import sys
import cv2
import numpy as np
import matplotlib.pyplot as plt

sys.path.insert(0, os.path.abspath('..'))
from src.video.reader import VideoReader
from src.enhancement.manager import EnhancementManager
```

## 1. Initialize Enhancement Manager

```python
manager = EnhancementManager()
sample_video = '../videos/VIRAT_S_010205_04_000545_000576.mp4'
reader = VideoReader(sample_video)
ret, frame = reader.read_frame()
reader.release()

print(f"Loaded frame with shape: {frame.shape}")
```

## 2. Multi-Model Enhancement Comparison

```python
enh_zero_dce, _ = manager.process_frame(frame, mode='zero_dce')
enh_rvrt, _ = manager.process_frame(frame, mode='rvrt')
enh_realesrgan, _ = manager.process_frame(frame, mode='realesrgan')
enh_auto, report = manager.process_frame(frame, mode='auto')

fig, axes = plt.subplots(1, 4, figsize=(24, 6))

axes[0].imshow(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
axes[0].set_title('Raw CCTV Input', fontsize=14)
axes[0].axis('off')

axes[1].imshow(cv2.cvtColor(enh_zero_dce, cv2.COLOR_BGR2RGB))
axes[1].set_title('Zero-DCE++ (Low Light)', fontsize=14)
axes[1].axis('off')

axes[2].imshow(cv2.cvtColor(enh_rvrt, cv2.COLOR_BGR2RGB))
axes[2].set_title('RVRT (Deblurred)', fontsize=14)
axes[2].axis('off')

axes[3].imshow(cv2.cvtColor(enh_auto, cv2.COLOR_BGR2RGB))
axes[3].set_title('Adaptive Pipeline Output', fontsize=14, color='darkgreen', fontweight='bold')
axes[3].axis('off')

plt.tight_layout()
plt.show()
```
